## Iterative Prompt Refinement

**Baseline prompt:** Analyze the following sentence and classify it as one of the six core emotions (happiness, sadness, anger, surprise, fear, disgust) or neutral. Answer with only one of the core emotions I mention in this prompt.

**Baseline prompt + few-shot prompting:** Analyze the following sentence and classify it as one of the six core emotions. Emotions: happiness, sadness, anger, surprise, fear, disgust or neutral. Answer in the following format: Sentence text: Emotion. When assigning emotion choose only one from the specified in the Emotions. Do not use any other emotions other than those mentioned.

**Refined prompt:** You are an expert psychologist and linguist trained to analyze emotions from text using both psychological theory and linguistic patterns. Use the following refined definitions and cues to identify the dominant emotion conveyed in a sentence. Happiness: A positive emotional state expressed through words or tone indicating pleasure, joy, amusement, satisfaction, love, gratitude, or excitement. Different from disgust: surprise focuses on unexpectedness, not rejection or revulsion. Only use these core emotions: happiness, sadness, anger, surprise, fear, disgust or neutral. Do not use any other emotions other than those mentioned
