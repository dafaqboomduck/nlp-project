```python
prompt = (
    "You are an expert emotion classifier. Analyze the sentence and respond with ONLY ONE lowercase emotion word. "
    "EMOTION DEFINITIONS: happiness is positive emotions like joy, pleasure, satisfaction, love, gratitude, excitement, pride, amusement, relief, or contentment, often shown by positive adjectives, affection, enjoyment, or laughter. "
    "Sadness is feelings of sorrow, grief, disappointment, loss, loneliness, melancholy, regret, or hopelessness, with cues like crying, tears, missing someone, loss, being alone, disappointment, depression, wishing, being sorry, or emotional hurt. "
    "Anger is irritation, frustration, rage, annoyance, resentment, or hostility from perceived injustice or obstruction, with cues like profanity, yelling, blame, complaints, 'I hate', aggressive tone, demands, or criticism. "
    "Fear is anxiety, worry, nervousness, dread, panic, or concern about threats or danger (real or imagined), with cues like scared, afraid, worried, anxious, nervous, 'what if', concerned, terrified, or panic. "
    "Disgust is revulsion, contempt, distaste, or aversion to something offensive or unpleasant, with cues like gross, disgusting, awful, terrible (quality), hate (food/object), revolting, nasty, or repulsive. "
    "Surprise is a reaction to unexpected events, positive or negative, focusing on unexpectedness, with cues like wow, really?, what?, can't believe, suddenly, unexpected, shocking, or amazing (unexpected context). "
    "Neutral means no clear emotion, factual statements, questions without emotional tone, observations, or matter-of-fact, with cues like simple statements, schedules, descriptions, or straightforward questions. "
    "CRITICAL RULES: Don't default to anger for all negative content. Sadness involves loss or sorrow, NOT just negativity. Fear involves threat or worry about the future, NOT past grievances. Disgust is about revulsion to specific things, NOT general negativity. Surprise requires unexpectedness, NOT just intensity. Use neutral when no emotion is clearly dominant."
)
```

```python
def extract_emotion(model_response):
    """Extract only the emotion from the model's response and convert to lowercase"""
    # Split by colon and take the last part
    if ':' in model_response:
        emotion = model_response.split(':')[-1].strip()
    else:
        # If no colon, try to extract after "Emotion"
        if 'Emotion' in model_response:
            emotion = model_response.split('Emotion')[-1].strip()
        else:
            emotion = model_response.strip()
    
    # Remove any remaining quotes, newlines, or extra characters
    emotion = emotion.replace('"', '').replace('\n', '').replace('\\n', '').strip().rstrip('.')
    
    # Convert to lowercase
    emotion = emotion.lower()
    
    return emotion


def chat_with_model_baseline_refined(token, sentence):
    url = 'http://194.171.191.228:30080/api/chat/completions'
    headers = {
        'Authorization': f'Bearer {token}',
        'Content-Type': 'application/json'
    }
    
    data = {
      "model": "meta-llama/Llama-3.3-70B-Instruct",
      "messages": [
            {"role": "system", "content": prompt},
            {"role": "user", "content": f'Classify this sentence (even if short): "{sentence}". You MUST respond with ONLY ONE WORD - the lower cased emotion name. USE ONLY ONE these core emotions: happiness, sadness, anger, surprise, fear, disgust or neutral. ALWAYS answer in the following format: Sentence : Emotion'}
      ],
      "context_length": 2048
    }
    response = requests.post(url, headers=headers, json=data)
    return response.json()


output = []

for idx, row in company_data.iterrows():
    sentence = row['Sentence']
    print(f"Processing row {idx}: {sentence[:50]}...")  
    
    json_response = chat_with_model_baseline_refined(token, sentence)
    content = json_response['choices'][0]['message']['content']
    
    # Extract only the emotion
    emotion = extract_emotion(content)
    
    output.append({'Sentence': sentence, 'Model_Emotion': emotion})

results_df = pd.DataFrame(output)
print(results_df.head())
```

```python
import sklearn
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
```

```python
df_comparison = results_df.copy()

company_data["LLM_Emotions"] = df_comparison["Model_Emotion"]

allowed_labels = {"happiness", "sadness", "anger", "surprise", "fear", "disgust", "neutral"}
company_data = company_data[company_data["LLM_Emotions"].isin(allowed_labels)].reset_index(drop=True)

print(df_comparison["Model_Emotion"].value_counts())

accuracy = accuracy_score(company_data['Emotion_core'], company_data['LLM_Emotions'])
print(f"Accuracy: {accuracy:.4f}")

print(classification_report(company_data['Emotion_core'], company_data['LLM_Emotions']))

print(confusion_matrix(company_data['Emotion_core'], company_data['LLM_Emotions']))
```
